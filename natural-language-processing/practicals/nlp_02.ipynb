# **NLP Practical 02**

Design a rule based chatbot.

```python
# Import the Chat class and default reflections dictionary from NLTK's chat utility module.
# 'Chat' is a engine for rule-based chatbot matching using regular expressions.
# 'reflections' is a dictionary that maps first-person pronouns to second-person (e.g., "I am" -> "you are").
from nltk.chat.util import Chat, reflections

# 'pairs' is a list of lists (pattern-response pairs) used for pattern matching.
# Each inner list contains:
#   1. A regular expression (regex) pattern string to match user input.
#   2. A list of response strings to reply with when the pattern matches.
pairs = [
    [
        # Pattern matches greetings like "hi", "hello", or "hey".
        # 'r' denotes a raw string so backslashes aren't treated as escape sequences.
        r"hi|hello|hey",
        ["Hello! Welcome to FIFA Shoe Store. How can I help you today?"]
    ],
    [
        # Pattern matches any input containing "shoe" or "shoes".
        # '(.*)' matches any sequence of characters before "shoe".
        # '(s)?' optionally matches the letter 's' at the end.
        r"(.*)shoe(s)?",
        ["We have Sports Shoes, Running Shoes, Sneakers, Formal Shoes and Casual Shoes."]
    ],
    [
        r"(.*)premium(.*)",
        ["Our Premium Shoes start from ₹1999."]
    ],
    [
        r"(.*)running(.*)",
        ["Running Shoes are available from Nike, Adidas and Puma."]
    ],
    [
        r"(.*)formal(.*)",
        ["We have premium formal shoes for office and business use."]
    ],
    [
        r"(.*)casual(.*)",
        ["Casual shoes are available in different colours and sizes."]
    ],
    [
        r"(.*)size(.*)",
        ["Available sizes are 6, 7, 8, 9, 10, 11 and 12."]
    ],
    [
        r"(.*)colour|color(.*)",
        ["Available colours are Black, White, Blue, Brown and Grey."]
    ],
    [
        r"(.*)price(.*)",
        ["Our shoe prices range from ₹999 to ₹6999."]
    ],
    [
        r"(.*)brand(.*)",
        ["We sell Nike, Adidas, Puma, Reebok, Woodland and Bata."]
    ],
    [
        r"(.*)discount(.*)",
        ["Flat 20 percent discount is available on selected shoes."]
    ],
    [
        r"(.*)payment(.*)",
        ["We accept Cash, UPI, Debit Card and Credit Card."]
    ],
    [
        r"(.*)delivery(.*)",
        ["Delivery takes 3 to 5 working days."]
    ],
    [
        r"(.*)return(.*)",
        ["Products can be returned within 7 days if unused."]
    ],
    [
        r"(.*)location(.*)",
        ["Our showroom is located near GSFC University, Vadodara."]
    ],
    [
        r"(.*)timing(.*)",
        ["Our store is open from 10 AM to 9 PM every day."]
    ],
    [
        r"(.*)contact(.*)",
        ["You can contact us at +91-9876543210."]
    ],
    [
        r"thank you|thanks",
        ["You're welcome! Happy Shopping."]
    ],
    [
        # Matches exit command.
        r"bye",
        ["Thank you for visiting FIFA Shoe Store. Have a great day!"]
    ],
    [
        # Fallback/Default rule: '.*' matches ANY input that failed to match previous patterns.
        # NLTK processes patterns sequentially top-to-bottom, so this acts as the catch-all.
        r"(.*)",
        ["Sorry, I didn't understand. Please ask about shoes, price, size, colour, brands or delivery."]
    ]
]

def shoebot():
    """
    Main function to run the FIFA Shoe Store Chatbot interaction loop.
    """
    # Display header banners for UI layout.
    print("=" * 50)
    print("        FIFA SHOE STORE CHATBOT")
    print("=" * 50)
    print("Type 'bye' to exit.\n")

    # Initialize the NLTK Chat object passing defined pattern pairs and pronoun reflections.
    chat = Chat(pairs, reflections)

    # Infinite loop to keep taking user input until 'bye' is received.
    while True:
        # Prompt user for input text.
        user = input("User: ")

        # Get the bot response using NLTK's regex matching logic.
        response = chat.respond(user)

        # Output the response to console.
        print("Chatbot:", response)

        # Check if user wants to exit (case-insensitive check).
        if user.lower() == "bye":
            break  # Terminate loop

# Entry point check: Ensures the bot runs directly when this script is executed.
if __name__ == "__main__":
    shoebot()
```
