```python
# Imports
import nltk
import re
import matplotlib.pyplot as plt
from collections import Counter

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# Open and read the text file
with open("data.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)

print("Original Text")
print(text)

# Tokenize the original corpus
sentences = sent_tokenize(text)

print("\nSentence Tokenization:\n")

for i, sentence in enumerate(sentences, 1):
    print(i, sentence)

words = word_tokenize(text)

print("\nWord Tokenization:\n")
print(words)
```

```python
# Stopword removal
stop_words = set(stopwords.words("english"))

filtered_words = []

for word in words:
    if word.lower().isalpha() and word.lower() not in stop_words:
        filtered_words.append(word.lower())

print("\nWords after Removing Stopwords:\n")
print(filtered_words)
```

```python
# Stemming and Lemmatization
stemmer = PorterStemmer()

stemmed_words = []

for word in filtered_words:
    stemmed_words.append(stemmer.stem(word))

print("\nStemmed Words:\n")
print(stemmed_words)

lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(tag):

    if tag.startswith('J'):
        return wordnet.ADJ

    elif tag.startswith('V'):
        return wordnet.VERB

    elif tag.startswith('N'):
        return wordnet.NOUN

    elif tag.startswith('R'):
        return wordnet.ADV

    else:
        return wordnet.NOUN

pos_tags = nltk.pos_tag(filtered_words)

lemmatized_words = []

for word, tag in pos_tags:

    lemma = lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    lemmatized_words.append(lemma)

print("\nLemmatized Words with POS:\n")
print(lemmatized_words)

print("\nSynonyms and Antonyms\n")

search_words = ["learning", "machine", "data", "students", "future"]

corpus = [w.lower() for w in filtered_words]

for word in search_words:

    if word in corpus:

        synonyms = set()
        antonyms = set()

        for syn in wordnet.synsets(word):

            for l in syn.lemmas():

                synonyms.add(l.name())

                if l.antonyms():

                    antonyms.add(l.antonyms()[0].name())

        print("\nWord :", word)

        print("Synonyms :", list(synonyms)[:5])

        if antonyms:
            print("Antonyms :", list(antonyms)[:5])
        else:
            print("Antonyms : Not Found")

    else:
        print(word, "-> Out of Corpus")

print("\nPhone Number Validation\n")

pattern = r'^\+91-[98]\d{9}$'

numbers = [
    "+91-9876543210",
    "+91-8123456789",
    "+91-7123456789"
]

for number in numbers:

    if re.fullmatch(pattern, number):
        print(number, "Valid")

    else:
        print(number, "Invalid")

print("\nSearch and Replace\n")

search_pattern = "machine"

if re.search(search_pattern, text, re.IGNORECASE):
    print("Pattern Found")

new_text = re.sub(search_pattern,
                  "AI",
                  text,
                  flags=re.IGNORECASE)

print("\nAfter Replacement:\n")
print(new_text)

print("\nWords Starting with Vowel\n")

vowel_words = re.findall(r'\b[AEIOUaeiou]\w*', text)

print(vowel_words)

#Plot frequency of words
print("\nPlotting Word Frequency...")

frequency = Counter(filtered_words)

x = list(frequency.keys())
y = list(frequency.values())

plt.figure(figsize=(14,6))
plt.bar(x, y)

plt.xticks(rotation=90)

# Word frequency after removing stopwords
plt.title("Word Frequency After Removing Stopwords")
plt.xlabel("Words")
plt.ylabel("Frequency")

plt.tight_layout()

plt.show()
```

```python
print("\nPOS Tagging\n")

tags = nltk.pos_tag(word_tokenize(text))

for tag in tags:
    print(tag)

nouns = []
verbs = []

for word, tag in tags:

    if tag.startswith("NN"):
        nouns.append(word)

    elif tag.startswith("VB"):
        verbs.append(word)

print("\nNouns:\n")
print(nouns)

print("\nVerbs:\n")
print(verbs)

adjectives = []

for word, tag in tags:

    if tag.startswith("JJ"):

        root = lemmatizer.lemmatize(word, wordnet.ADJ)
        adjectives.append(root)

print("\nNumber of Adjectives :", len(adjectives))

print("\nRoot Words of Adjectives:\n")
print(adjectives)
```
