# **NLP Practical 04**

Select any research paper
1. Find the key words from the abstract of the research paper, and then compare with the actual keywords.
2. Summarize the research paper in 1/10th of the number of words (Eg. if the research paper has 400 words then summarize into 40 words) .

```python
import nltk
import re
from collections import Counter
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
abstract = """
Keyword extraction is the task of identifying important terms or phrase that
are most representative of the source document. Although the process of
automatic extraction of keywords from title is an old method, it was mainly for
extraction from a single web document. Our approach differs from previous
research works on keyword extraction in several aspects. For those who are
non-expert of the scientific fields, understating scientific research trends is
difficult. The purpose of this study is to develop an automatic method of
obtaining overviews of a scientific field for non-experts by capturing research
trends. This empirical study excavates significant term extraction using
Natural Language Processing (NLP) tools. More than 15000 titles saved in a
.csv file was our dataset and scripts written in Python were our process to
compare how far significant terms of scientific title corpus are similar or
different to the terms available in the abstract of that same scientific
article corpus. A light-weight unsupervised title extractor, Yet Another
Keyword Extractor (YAKE) was used to extract the results. Based on our
analysis, it can be concluded that these algorithms can be used for other
fields too by the non-experts of that subject field to perform automatic
extraction of significant words and understanding trends. Our algorithm could
be a solution to reduce the labour-intensive manual indexing process.
"""
```

```python
actual_keywords = [
    "data mining",
    "title extraction",
    "natural language processing",
    "yake",
    "nltk",
    "keyword extraction-nlp"
]

words = word_tokenize(abstract.lower())

print("=" * 70)
print("KEYWORD EXTRACTION FROM RESEARCH PAPER ABSTRACT")
print("=" * 70)

print("\nTotal tokens:", len(words))


stop_words = set(stopwords.words("english"))

clean_words = []

for word in words:

    # Keep alphabetic words only
    if word.isalpha() and word not in stop_words:
        clean_words.append(word)


print("\nNumber of words after stop-word removal:",
      len(clean_words))


word_frequency = Counter(clean_words)


print("\n" + "=" * 70)
print("MOST FREQUENT WORDS")
print("=" * 70)

for word, count in word_frequency.most_common(15):

    print(f"{word:<25} : {count}")


extracted_keywords = [
    word for word, count in word_frequency.most_common(10)
]


print("\n" + "=" * 70)
print("EXTRACTED KEYWORDS")
print("=" * 70)

for i, word in enumerate(extracted_keywords, 1):

    print(f"{i}. {word}")


print("\n" + "=" * 70)
print("COMPARISON WITH ACTUAL KEYWORDS")
print("=" * 70)


actual_keywords_lower = [
    keyword.lower()
    for keyword in actual_keywords
]


matched_keywords = []
unmatched_keywords = []


for keyword in actual_keywords_lower:

    keyword_words = keyword.split()

    if all(word in clean_words for word in keyword_words):

        matched_keywords.append(keyword)

    else:

        unmatched_keywords.append(keyword)


print("\nActual Keywords from Paper:")

for keyword in actual_keywords:
    print("-", keyword)


print("\nKeywords Found in Abstract:")

for keyword in matched_keywords:
    print("-", keyword)


print("\nKeywords Not Completely Found:")

for keyword in unmatched_keywords:
    print("-", keyword)


print("\n" + "=" * 70)
print("IMPORTANT TWO-WORD PHRASES (BIGRAMS)")
print("=" * 70)


bigrams = list(nltk.bigrams(clean_words))

bigram_frequency = Counter(bigrams)


for (word1, word2), count in bigram_frequency.most_common(15):

    phrase = word1 + " " + word2

    print(f"{phrase:<40} : {count}")


print("\n" + "=" * 70)
print("FINAL RESULT")
print("=" * 70)

print("\nThe abstract was tokenized and stop words were removed.")
print("The remaining words were counted according to their frequency.")
print("The most frequent words were selected as candidate keywords.")

print("\nAutomatically Extracted Keywords:")

for keyword in extracted_keywords:

    print("-", keyword)

print("\nActual Keywords:")

for keyword in actual_keywords:

    print("-", keyword)

print("\nThe extracted keywords can be compared with the")
print("author-provided keywords to determine their similarity.")
```
